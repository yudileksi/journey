```python
import requests

url = requests.get("https://jsonplaceholder.typicode.com/posts/1")
posts = url.json()

print(url.status_code)

for post in posts:
    print(post)
```

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users/1")

if response.status_code == 200:
    user = response.json()
    print("Name:", user["name"])
    print("Email:", user["email"])
    print("City:", user["address"]["city"])
else:
    print("Something went wrong:", response.status_code)
```

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users")

users = response.json()

print(response.status_code)

for user in users:
    print(user["name"], "-", user["email"])
```

```python
import requests

params = {
    "latitude": -7.9797,    # Malang's latitude
    "longitude": 112.6304,  # Malang's longitude
    "current_weather": True
}

response = requests.get("https://api.open-meteo.com/v1/forecast", params=params)

if response.status_code == 200:
    data = response.json()
    weather = data["current_weather"]
    
    print("Temperature:", weather["temperature"], "°C")
    print("Wind Speed:", weather["windspeed"], "km/h")
    print("Weather Code:", weather["weathercode"])
    print(response.url)
else:
    print("Something Went Wrong", response.status_code)
```

```python
import requests

params = {
    "latitude": -7.9797,    # Malang's latitude
    "longitude": 112.6304,  # Malang's longitude
    "current_weather": True,
    "hourly": "temperature_2m,precipitation",
    "timezone": "Asia/Jakarta"
}

response = requests.get("https://api.open-meteo.com/v1/forecast", params=params)
data = response.json()
   
print("Current Weather:")
print(data["current_weather"])

print("\nHourly temperatures (first 5 hours):")
print(data["hourly"]["temperature_2m"][:5])
```

```python
import requests

response = requests.get("https://restcountries.com/v3.1/name/indonesia")

if response.status_code == 200:
    country = response.json()[0]  # returns a list, take first item
    
    print(country)
else:
    print("Something Went Wrong", response.status_code)
```

```python
import requests

response = requests.get("https://restcountries.com/v3.1/name/india")

if response.status_code == 200:
    country = response.json()[0]  # returns a list, take first item
    
    print("Country:", country["name"]["common"])
    print("Capital:", country["capital"][0])
    print("Population:", country["population"])
    print("Region:", country["region"])
    print("Currency:", list(country["currencies"].keys())[0])
else:
    print("Something Went Wrong", response.status_code)
```
