```python
import os
from groq import Groq
from dotenv import load_dotenv
import tiktoken

load_dotenv()
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

encoder = tiktoken.get_encoding("cl100k_base")

conversation_history = [
    {
        "role": "system",
        "content": """You are Kopi, a friendly AI assistant for a coffe shop in Malang, Indonesia. You know everything about coffee and help customers choose their perfect drink.
        Keep responses warm, conversational, and under 100 words"""
    }
]

def chat_stream(user_message):
    conversation_history.append({
        "role": "user",
        "content": user_message
    })

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=conversation_history,
        temperature=0.7,
        max_tokens=200,
        stream=True
    )

    print("Kopi: ", end="", flush=True)

    full_response = ""

    for chunk in response:

        delta = chunk.choices[0].delta.content
        if delta:
            print(delta, end="", flush=True)
            full_response += delta

    if chunk.usage is not None:
            print(f"\n\n📊 [Tokens Used -> Prompt: {chunk.usage.prompt_tokens} | Completion: {chunk.usage.completion_tokens} | Total: {chunk.usage.total_tokens}]")

    print()

    conversation_history.append({
        "role": "assistant",
        "content": full_response
    })

    history_text = "\n".join([f"{msg['role']}: {msg['content']}" for msg in conversation_history])
    tokens = encoder.encode(history_text)

    print(f"\n[Current Token Usage: {len(tokens)}]")

print("☕ Kopi Coffee Assistant (type 'quit' to exit)")
print("=" * 40)

while True:
    user_input= input("\nYou: ")

    if user_input.lower() == "quit":
        print("Goodbye! Come back for more coffee! ☕")
        break

    chat_stream(user_input)
```
